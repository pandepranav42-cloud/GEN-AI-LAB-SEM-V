```python
from transformers import pipeline
import torch
generator = pipeline(
    "text-generation",
    model="distilgpt2"
)

prompt = "Artificial Intelligence is"

result = generator(
    prompt,
    max_new_tokens=30,
    do_sample=False
)

print(result[0]["generated_text"])
```
